**SECCIÓN 1: CARGA E INSPECCIÓN INICIAL**

Instrucción: Carga el dataset del Titanic desde la siguiente URL pública usando pandas:

URL: "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

```python
import pandas as pd
```

```python
# Dataset
df=pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

# Mostrar las primeras 5 filas
print(df.head(5))

# Mostrar tipos de datos, memoria y valores nulos.
print(df.info())
```

**SECCIÓN 2: SELECCIÓN Y FILTRADO DE DATOS**

Instrucción: Extrae la columna 'Age' como una Serie de Pandas y calcula su promedio (mean).

Imprime el resultado con un mensaje descriptivo.

Instrucción: Crea un filtro (máscara booleana) para obtener únicamente a los pasajeros que viajaban en Primera Clase (Pclass == 1) y cuya tarifa (Fare) haya sido mayor a $100 USD.

Almacena el resultado en un nuevo DataFrame e imprime cuántos pasajeros cumplen con esta condición.

```python
# Extracción de columna age, promedio y mensaje descriptivo
age= df['Age']
p_age= age.mean()
print("Edad promedio" , p_age)

# Filtro de pasajeros, datafarme e impresión
filtro=(df['Pclass']==1)&(df['Fare']>100)
Pasajeros=df[filtro]
print=("Pasajeros que cumplen la condición:" , Pasajeros)
```

**SECCIÓN 3: DETECCIÓN Y TRATAMIENTO DE VALORES NULOS**



Instrucción: Muestra la cantidad total de valores nulos (NaN) por cada columna del DataFrame.

Instrucción: Rellena los valores nulos de la columna 'Age' utilizando la mediana de esa misma columna.

Asegúrate de modificar el DataFrame original o reasignar la columna.

Comprueba que ya no queden nulos en la columna 'Age'.

```python
# Mostrar valores nulos
print(df.isnull().sum())

# Rellenar valores nulos de age
mediana_age = df['Age'].median()

# Asegúrate de modificar el DataFrame original o reasignar la columna.
df['Age'].fillna(mediana_age)

# Comprobar que ya no queden nulos en la columna 'Age'
print(df.isnull().sum())


```

**SECCIÓN 4: CREACIÓN DE NUEVAS CARACTERÍSTICAS (FEATURE ENGINEERING)**

Instrucción: Crea una nueva columna llamada 'FamilySize' que sume el número de hermanos/cónyuges ('SibSp'), el número de padres/hijos ('Parch') y el 1 (que representa al pasajero mismo).

Muestra las primeras 3 filas de las columnas: 'Name', 'SibSp', 'Parch' y 'FamilySize'.


```python
# Suma de columnas
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
# Muestra las primeras 3 filas de las columnas: 'Name', 'SibSp', 'Parch' y 'FamilySize'.
print(df[['Name', 'SibSp', 'Parch', 'FamilySize']].head(3))
```

**SECCIÓN 5: AGRUPAMIENTOS Y AGREGACIONES (GROUPBY)**

Instrucción: Agrupa el DataFrame por género ('Sex') y calcula la tasa de supervivencia promedio (el promedio de la columna 'Survived'), multiplícalo por 100 para verlo en porcentaje

Instrucción: Agrupa el DataFrame por clase de pasajero ('Pclass') y aplica múltiples funciones de agregación sobre la tarifa ('Fare'): calcula la media, el valor mínimo y el valor máximo.

```python
# Tasa de supervivencia promedio
df.groupby("Sex")["Survived"].mean()*100

# Media, valor mínimo y valor máximo Pclass y Fare
df.groupby("Pclass")["Fare"].agg(["mean", "min", "max"])

```
