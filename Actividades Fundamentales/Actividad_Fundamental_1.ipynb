### Ejercicio 1: Evaluador de Rendimiento de Modelos
Se desarrolla una función que solicita al usuario la exactitud de un modelo de inteligencia artificial mediante `input()`. Utilizando estructuras condicionales `if`, `elif` y `else`, se clasifica su rendimiento en cuatro categorías: excelente, aceptable, requiere ajuste o crítico. También se verifica que el valor ingresado esté dentro del rango válido de 0 a 1.

```python
# Ejercicio 1: Evaluador de Rendimiento de Modelos

def evaluar_modelo():
    exactitud = float(input("Ingresa la exactitud del modelo: "))

    if exactitud < 0 or exactitud > 1:
        return "Exactitud no válida"
    elif exactitud >= 0.90:
        return "Excelente"
    elif exactitud >= 0.75:
        return "Aceptable"
    elif exactitud >= 0.50:
        return "Requiere ajuste"
    else:
        return "Crítico / No apto"

# Espacio de prueba
print(evaluar_modelo())
```

### Ejercicio 2: Procesamiento de Hiperparámetros y Lotes

Se utiliza una tupla para almacenar una configuración fija de dimensiones de entrada y salida. Posteriormente, una función recibe una lista de tasas de aprendizaje y utiliza un ciclo `for` para multiplicar cada valor por un factor de escala de 2.0. Los resultados se almacenan en una nueva lista y se muestran junto con la configuración original.

```python
# Ejercicio 2: Procesamiento de Hiperparámetros y Lotes

CONFIGURACION_BASE = (512, 10)

def procesar_tasas_aprendizaje(tasas):
    tasas_escaladas = []

    for x in tasas:
        tasas_escaladas.append(float(x) * 2)

    return tasas_escaladas, CONFIGURACION_BASE

# Espacio de prueba
tasas = input("Insertar las tasas separadas por comas: ").split(",")

print(procesar_tasas_aprendizaje(tasas))
```

### Ejercicio 3: Simulador de Convergencia por Épocas

**Nombre:** [Nombre completo]  
**Matrícula:** [Matrícula]

Se implementa una función que solicita al usuario el número máximo de épocas de entrenamiento. Mediante un ciclo se simula la disminución progresiva de la pérdida (`loss`) en cada época. Se utiliza una condición `if` para comprobar si la pérdida es inferior a 0.05 y, en caso afirmativo, detener anticipadamente el entrenamiento al alcanzar la convergencia.

```python
# Ejercicio 3: Simulador de Convergencia por Épocas

def simular_entrenamiento():
    max_epocas = int(input("Ingresa el máximo de épocas: "))

    for epoca in range(1, max_epocas + 1):
        loss = 1 / epoca

        print("Época:", epoca, "| Loss:", round(loss, 4))

        if loss < 0.05:
            print("El modelo ha convergido en la época", epoca)
            break

simular_entrenamiento()
```
