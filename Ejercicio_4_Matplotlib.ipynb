## **Análisis Exploratorio de Datos (EDA) del Titanic**

**Objetivo**. Descargar un dataset real directamente desde internet usando Pandas, realizar una limpieza básica y construir un dashboard para analizar qué factores (edad, clase, tarifa) influyeron en la supervivencia. Elimina las filas que no tengan el dato de la edad ()

*Parte 1: Preparación y Dashboard Básico*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Dataset
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

# Elimina las filas que no tengan el dato de la edad (Age) utilizando .dropna() para evitar errores en los gráficos.
df = df.dropna(subset=['Age'])
```

```python
# Crea una figura con que contenga una cuadrícula de 2x2.
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Genera los siguientes gráficos en cada cuadrante usando las herramientas vistas en clase:
# Arriba Izquierda (Histograma) para mostrar la distribución de las edades de todos los pasajeros.
axes[0, 0].hist(df["Age"], bins=20)
axes[0, 0].set_title("Distribución de edades")
axes[0, 0].set_xlabel("Edad")
axes[0, 0].set_ylabel("Número de pasajeros")

# Arriba Derecha (Gráfico de Barras) para mostrar cuántos pasajeros había en cada clase.
clases = df["Pclass"].value_counts().sort_index()
axes[0, 1].bar(clases.index, clases.values)
axes[0, 1].set_title("Pasajeros por clase")
axes[0, 1].set_xlabel("Clase")
axes[0, 1].set_ylabel("Número de pasajeros")

# Abajo Izquierda (Dispersión) para graficar la Edad (eje X) o (eje Y) contra la Tarifa.
axes[1, 0].scatter(df["Age"], df["Fare"])
axes[1, 0].set_title("Edad vs Tarifa")
axes[1, 0].set_xlabel("Edad")
axes[1, 0].set_ylabel("Tarifa (Fare)")

# Abajo Derecha (Gráfico de Línea) para mostrar la tarifa promedio pagada según la edad.
fare_por_edad = df.groupby("Age")["Fare"].mean()
axes[1, 1].plot(fare_por_edad.index, fare_por_edad.values)
axes[1, 1].set_title("Tarifa promedio según edad")
axes[1, 1].set_xlabel("Edad")
axes[1, 1].set_ylabel("Tarifa promedio")

plt.tight_layout()
plt.show()
```

*Parte 2: Selección de características*

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Mapa de calor
# Dataset
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
#Matriz correlación
matrizcorr = df.corr(numeric_only=True)

fig, ax = plt.subplots(figsize=(10, 8))
plt.imshow(matrizcorr, cmap="coolwarm", vmin=-1, vmax=1)
plt.colorbar()
plt.xticks(range(len(matrizcorr.columns)), matrizcorr.columns, rotation=90)
plt.yticks(range(len(matrizcorr.columns)), matrizcorr.columns)
plt.title("Mapa de calor de Matriz de correlación")
plt.show()
```
