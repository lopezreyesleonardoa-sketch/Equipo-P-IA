```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")
df = df.dropna(subset=["Age"])
fig, axs = plt.subplots(2,2)
axs[0, 0].hist(df["Age"])
clases = df["Pclass"].value_counts()
axs[0, 1].bar(clases.index, clases.values)
axs[1, 0].scatter(df["Age"], df["Fare"])
tarifa_por_edad = df.groupby("Age")["Fare"].mean()
axs[1, 1].plot(tarifa_por_edad.index, tarifa_por_edad.values)
plt.tight_layout()
plt.show()

columnas_numericas = df.select_dtypes(include="number")
correlacion = columnas_numericas.corr()

plt.figure(figsize=(10, 8))
plt.imshow(correlacion, cmap="coolwarm", aspect="auto")
plt.colorbar()

plt.xticks(
    range(len(correlacion.columns)),
    correlacion.columns,
    rotation=45
)

plt.yticks(
    range(len(correlacion.columns)),
    correlacion.columns
)

plt.title("Mapa de calor de correlaciones")
plt.tight_layout()
plt.show()
```
