```python

import pandas as pd

# ==========================================
# SECCIÓN 1: CARGA E INSPECCIÓN INICIAL
# ==========================================
# Instrucción: Carga el dataset del Titanic desde la siguiente URL pública usando pandas:
# URL: "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
# Asigna el resultado a un DataFrame llamado 'df'.

# Escribe tu código aquí:
df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

# Instrucción: Muestra las primeras 5 filas del DataFrame para analizar su estructura y columnas.
df.head()
# Instrucción: Imprime un resumen general del DataFrame (tipos de datos, memoria y nulos) usando el método adecuado.
df.info()
# ==========================================
# SECCIÓN 2: SELECCIÓN Y FILTRADO DE DATOS
# ==========================================
# Instrucción: Extrae la columna 'Age' como una Serie de Pandas y calcula su promedio (mean).
# Imprime el resultado con un mensaje descriptivo.
df["Age"]
promedio_edad = df["Age"].mean()
print("Promedio de edad:", promedio_edad)
# Instrucción: Crea un filtro (máscara booleana) para obtener únicamente a los pasajeros 
# que viajaban en Primera Clase (Pclass == 1) y cuya tarifa (Fare) haya sido mayor a $100 USD.
# Almacena el resultado en un nuevo DataFrame e imprime cuántos pasajeros cumplen con esta condición.
df["Pclass"] == 1
df["Fare"]>100
mascara = (df["Pclass"] == 1) & (df["Fare"] > 100)
pasajeros_filtrados = df[mascara]
print("Pasajeros filtrados" , str(len(pasajeros_filtrados)))
# ==========================================
# SECCIÓN 3: DETECCIÓN Y TRATAMIENTO DE VALORES NULOS
# ==========================================
# Instrucción: Muestra la cantidad total de valores nulos (NaN) por cada columna del DataFrame.
df.isnull().sum()
# Instrucción: Rellena los valores nulos de la columna 'Age' utilizando la mediana de esa misma columna.
# Asegúrate de modificar el DataFrame original o reasignar la columna.
mediana_edad = df["Age"].median()
df["Age"] = df["Age"].fillna(mediana_edad)
# Comprueba que ya no queden nulos en la columna 'Age'.
df["Age"].isnull().sum()
# ==========================================
# SECCIÓN 4: CREACIÓN DE NUEVAS CARACTERÍSTICAS (FEATURE ENGINEERING)
# ==========================================
# Instrucción: Crea una nueva columna llamada 'FamilySize' que sume el número de hermanos/cónyuges ('SibSp'),
# el número de padres/hijos ('Parch') y el 1 (que representa al pasajero mismo).
# Muestra las primeras 3 filas de las columnas: 'Name', 'SibSp', 'Parch' y 'FamilySize'.
df["FamilySize"]= df["SibSp"]+df["Parch"]+1
df[["Name", "SibSp", "Parch", "FamilySize"]].head(3)
# ==========================================
# SECCIÓN 5: AGRUPAMIENTOS Y AGREGACIONES (GROUPBY)
# ==========================================
# Instrucción: Agrupa el DataFrame por género ('Sex') y calcula la tasa de supervivencia promedio 
# (el promedio de la columna 'Survived'), multiplícalo por 100 para verlo en porcentaje.
df.groupby("Sex")["Survived"].mean()*100
# Instrucción: Agrupa el DataFrame por clase de pasajero ('Pclass') y aplica múltiples funciones 
# de agregación sobre la tarifa ('Fare'): calcula la media, el valor mínimo y el valor máximo.
df.groupby("Pclass")["Fare"].agg(["mean", "min", "max"])
```
